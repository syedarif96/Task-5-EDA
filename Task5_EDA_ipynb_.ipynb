````python

import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x='Survived')

plt.title('Passenger Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')

plt.show()
````

````python

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Age', bins=20, kde=True)

plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
````

````python

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Age')

plt.title('Boxplot of Passenger Ages')
plt.xlabel('Age')

plt.show()
````

````python

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived')

plt.title('Passenger Age vs Ticket Fare')
plt.xlabel('Age')
plt.ylabel('Ticket Fare')
plt.legend(title='Survived', labels=['Did Not Survive', 'Survived'])

plt.show()
````

````python

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Pclass', hue='Survived')

plt.title('Passenger Survival by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['Did Not Survive', 'Survived'])

plt.show()
````

````python

plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(include='number').corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Titanic Dataset')
plt.show()
````

**Observation – Correlation Heatmap**

The correlation heatmap displays the linear relationships between numerical variables in the Titanic dataset. Values closer to +1 indicate positive correlations, while values closer to -1 indicate negative correlations. Values near zero indicate weak linear relationships. The heatmap helps identify variables associated with survival, although correlation does not prove causation.


````python

sns.pairplot(
    df[['Survived', 'Pclass', 'Age', 'Fare']].dropna(),
    hue='Survived',
    diag_kind='hist'
)

plt.show()
````


**Observation – Pairplot of Numerical Variables**

The pairplot compares passenger survival, class, age, and fare across multiple plots. It helps explore relationships and differences between passengers who survived and those who did not. The plots may reveal patterns involving passenger class and fare, while the age distribution can show how the recorded ages vary. The pairplot is exploratory and does not establish causation.

# Final Summary of Findings

The Exploratory Data Analysis was performed on the Titanic passenger dataset using Python, Pandas, Matplotlib, and Seaborn.

**Key findings:**

1. **Dataset overview:** The dataset contains 891 passenger records and 12 columns.
2. **Missing values:** Age has 177 missing values, Cabin has 687, and Embarked has 2.
3. **Survival:** The dataset contains more passengers who did not survive than passengers who survived.
4. **Age distribution:** The histogram shows the distribution of recorded passenger ages, while the boxplot helps identify potential age outliers.
5. **Passenger class:** Survival outcomes differ across passenger classes, which can be explored using the class comparison chart.
6. **Gender:** The gender comparison reveals differences in survival outcomes between male and female passengers.
7. **Relationships:** The scatterplot, pairplot, and correlation heatmap help explore relationships among numerical variables.

**Conclusion:**

EDA helped reveal the structure, data quality, distributions, and relationships within the Titanic dataset. Visualizations and statistical summaries provided useful insights into passenger characteristics and survival patterns. This analysis demonstrates how Python can be used to explore and understand a dataset before further statistical analysis or modeling.


````python

from matplotlib.backends.backend_pdf import PdfPages
from IPython.display import FileLink, display
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

pdf_path = '/content/EDA_Report.pdf'

# Calculate findings from your actual dataset
total = len(df)
survived = int((df['Survived'] == 1).sum())
not_survived = int((df['Survived'] == 0).sum())
age_missing = int(df['Age'].isnull().sum())
cabin_missing = int(df['Cabin'].isnull().sum())
embarked_missing = int(df['Embarked'].isnull().sum())

with PdfPages(pdf_path) as pdf:

    # Page 1: Summary of findings
    fig = plt.figure(figsize=(8.27, 11.69))
    fig.suptitle('Exploratory Data Analysis (EDA)',
                 fontsize=20, fontweight='bold', y=0.95)

    summary = f"""
Titanic Passenger Dataset

Dataset overview
• Total passenger records: {total}
• Total columns: {df.shape[1]}
• Passengers who survived: {survived} ({survived/total:.1%})
• Passengers who did not survive: {not_survived}
  ({not_survived/total:.1%})

Missing values
• Age: {age_missing}
• Cabin: {cabin_missing}
• Embarked: {embarked_missing}

Key findings
• More passengers did not survive than survived.
• Age values vary across passengers.
• Survival outcomes differ by passenger class.
• Survival outcomes also differ by gender.
• The charts help explore numerical relationships
  and identify potential outliers.

Conclusion
Pandas was used for statistical exploration, while
Matplotlib and Seaborn were used to create visualizations.
The analysis highlights data quality, distributions,
and patterns in Titanic passenger survival.

Note: Findings describe associations in this dataset
and do not establish cause and effect.
"""
    fig.text(0.10, 0.84, summary, fontsize=12,
             va='top', linespacing=1.6)
    plt.axis('off')
    pdf.savefig(fig, bbox_inches='tight')
    plt.close(fig)

    # Page 2: Survival counts
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.countplot(data=df, x='Survived', ax=ax)
    ax.set_title('Passenger Survival Count')
    ax.set_xlabel('Survived (0 = No, 1 = Yes)')
    ax.set_ylabel('Number of Passengers')
    pdf.savefig(fig, bbox_inches='tight')
    plt.close(fig)

    # Page 3: Age distribution
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.histplot(data=df, x='Age', bins=20, kde=True, ax=ax)
    ax.set_title('Distribution of Passenger Ages')
    ax.set_xlabel('Age')
    ax.set_ylabel('Number of Passengers')
    pdf.savefig(fig, bbox_inches='tight')
    plt.close(fig)

    # Page 4: Survival by passenger class
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.countplot(data=df, x='Pclass', hue='Survived', ax=ax)
    ax.set_title('Passenger Survival by Class')
    ax.set_xlabel('Passenger Class')
    ax.set_ylabel('Number of Passengers')
    pdf.savefig(fig, bbox_inches='tight')
    plt.close(fig)

    # Page 5: Survival by gender
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.countplot(data=df, x='Sex', hue='Survived', ax=ax)
    ax.set_title('Passenger Survival by Gender')
    ax.set_xlabel('Gender')
    ax.set_ylabel('Number of Passengers')
    pdf.savefig(fig, bbox_inches='tight')
    plt.close(fig)

    # Page 6: Correlation heatmap
    numeric_corr = df.select_dtypes(
        include='number'
    ).corr()

    fig, ax = plt.subplots(figsize=(9, 7))
    sns.heatmap(numeric_corr, annot=True, cmap='coolwarm',
                fmt='.2f', ax=ax)
    ax.set_title('Correlation Heatmap')
    pdf.savefig(fig, bbox_inches='tight')
    plt.close(fig)

print("PDF report created successfully!")
display(FileLink(pdf_path))
````
